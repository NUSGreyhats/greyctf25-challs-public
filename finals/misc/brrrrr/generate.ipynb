```python
from scipy.io.wavfile import write
import numpy as np

def psk_modulate(data, carrier_freq, sample_rate):
    # Convert string to binary
    binary_data = ''.join(format(ord(char), '08b') for char in data) * 20

    baseband = [1 if x == '1' else -1 for x in binary_data]
    baseband = np.repeat(baseband, 48)

    # Generate carrier signal
    t = np.arange(0, len(baseband) / sample_rate, 1 / sample_rate)
    carrier = np.sin(2 * np.pi * carrier_freq * t)
    
    modulated_signal = np.zeros(1234 + len(carrier) + 4321)
    modulated_signal[1234:-4321] = baseband * carrier

    return modulated_signal + np.random.normal(0, 0.01, len(modulated_signal))

# Parameters
data = "grey{ph4s3_sh1ft_k3y1ng_3498th45789tgh}"
carrier_freq = 1000  # 1 kHz
sample_rate = 48000  # 48 kHz

# Generate PSK modulated signal
modulated_signal = psk_modulate(data, carrier_freq, sample_rate)

# Normalize signal to 16-bit range
modulated_signal = np.int16(modulated_signal / np.max(np.abs(modulated_signal)) * 4096)

# Save to audio file
write("psk_modulated.wav", sample_rate, modulated_signal)
```

```python
from scipy.signal import butter, lfilter
from scipy.io.wavfile import read

# Read the modulated signal from the audio file
sample_rate, modulated_signal = read("psk_modulated.wav")

def psk_demodulate(modulated_signal, carrier_freq, sample_rate):
    # Generate carrier signal for mixing down
    modulated_signal = modulated_signal[1234:-4321]

    t = np.arange(len(modulated_signal)) / sample_rate
    #carrier = np.exp(-1j * 2 * np.pi * carrier_freq * t)
    carrier = np.sin(2 * np.pi * carrier_freq * t)

    # Mix down to baseband
    baseband_signal = modulated_signal * carrier
    
    # Design low-pass filter
    #cutoff_freq = carrier_freq / 2  # Cutoff frequency
    #nyquist_rate = sample_rate / 2
    #normalized_cutoff = cutoff_freq / nyquist_rate
    #b, a = butter(5, normalized_cutoff, btype='low')
    
    # Apply low-pass filter
    #demodulated_signal = lfilter(b, a, baseband_signal)
    
    demodulated_signal = baseband_signal.reshape(-1, 48).sum(axis=-1)

    return (demodulated_signal > 60000).astype("int")

# Demodulate the signal
demodulated_signal = psk_demodulate(modulated_signal, carrier_freq, sample_rate)
```

```python
# Convert binary data to text
binary_string = ''.join(map(str, demodulated_signal))
decoded_text = ''.join(chr(int(binary_string[i:i+8], 2)) for i in range(0, len(binary_string), 8))
print(decoded_text)
```
